```python
employees = [
(1,"Saravana","IT",60000),
(2,"John","Sales",45000),
(3,"David","HR",50000)
]

columns=["EmpId","Name","Department","Salary"]

df=spark.createDataFrame(employees,columns)

'''display(df)
df.count()
df.show()
df.describe().show()'''

df.write.mode("overwrite").saveAsTable("employees")
```
